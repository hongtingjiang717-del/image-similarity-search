### 操作数据库，最常见的就是增删改查
### 什么是向量？就是一串有顺序的数字； 什么是嵌入？用向量表示文字、图片等数据； 什么是嵌入模型？负责把原始数据转换成向量的模型；
### 整个过程：
资料 到 嵌入模型  到  向量  到  存入chroma
问题  到  同一个嵌入模型  到  问题向量   chroma找相近的向量    返回向量

### 一个collection包含什么？
参数	含义	例子
ids	唯一编号，必须是字符串	"001"
documents	原始文本	"这里出售面包"
embeddings	文本或图片对应的向量	[0.12, 0.85, ...]
metadatas	附加标签，方便筛选	{"category": "food"}

只传documents，chroma就可以调用模型自动生成向量；模型负责提取特征，chroma负责存储和检索

操作	代码	含义
创建集合	client.create_collection(name="other_notes")	同名集合已存在时会报错
获取集合	client.get_collection(name="my_notes")	集合必须已存在
查看集合	client.list_collections()	看有哪些抽屉
查看数量	collection.count()	有多少条记录
随便看几条	collection.peek(limit=2)	检查数据
按 ID 获取	collection.get(ids=["1"])	准确找到指定记录
新增	collection.add(ids=["4"], documents=["Hello"])	添加新记录
修改	collection.update(ids=["4"], documents=["Hello world"])	修改已有记录
新增或修改	collection.upsert(ids=["4"], documents=["Hi"])	不存在则新增，存在则修改
删除记录	collection.delete(ids=["4"])	删除指定记录
删除集合	client.delete_collection(name="other_notes")	删除整个抽屉

```python
# 区分两种查询方式
# get：按编号或条件取数据，不按相似程度排序
collection.get(ids=["1"])

# query：在食物类别里，寻找意思最接近的两条
collection.query(
    query_texts=["I want some fruit."],
    where={"category": "food"},  #表示最多满足的筛选条件
    n_results=2  #表示最多返回2条
)
```
