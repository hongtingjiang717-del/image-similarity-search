```python
import chromadb
```

```python
# 1.创建Chroma客户端
client = chromadb.Client()
```

```python
# 2.创建集合
collection = client.create_collection(name = "test_collection")
```

```python
# 3.插入数据
collection.add(
    ids = ["1", "2", "3"],
    documents=[
        "This is the first document.",
        "This is the second document.",
        "This is the third document.",
    ]
)
```

```python
# 4.查看数据
collection.count()
```

```python
collection.peek(limit = 2)['embeddings']
```

```python
# 5.按相似度查询
collection.query(
    query_texts = ["This is the first document.", "This is the second document."]
)
```
