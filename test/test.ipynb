```python
from chromadb.experimental.density_relevance import dataset
from pyparsing import alphanums

img_name1 = "class01imag100.jpg"
img_name2 = "class01imag2.jpg"
print(img_name1>img_name2)
print(img_name1==img_name2)
```

```python
import re
img1 = re.split(r'([0-9]+)', img_name1)
img2 = re.split(r'([0-9]+)', img_name2)
print(img1)
print(img2) #这是一个列表
```

```python
print(img1>img2)
print(img1==img2)
```

```python
#定义一个转换函数，字符转成小写，如果是数字则提取数值
convert = lambda str:int(str) if str.isdigit() else str.lower()
```

```python
img1 = [convert(x) for x in img1]
print(img1)
img2 = [convert(x) for x in img2]
print(img2)
```

```python
print(img1>img2)
print(img1==img2)
```

```python
#定义一个排序key函数，按照字母-数字切分，得到列表作为key
alphanum = lambda str:[convert(x) for x in re.split(r'([0-9]+)',str)]
print(sorted(imgs,key=alphanum))
```
