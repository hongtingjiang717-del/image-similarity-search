## 对chromadb进行操作

```python
# 1.创建数据
collection = client.create_collection('my_collection')
```

```python
# 2.插入数据
collection.add(
    ids = ["1", "2", "3"],
    documents=[
        "This is the first document.",
        "This is the second document.",
        "This is the third document.",
    ]
)
```

```python
# 3.按相似度查询
collection.query(
    query_texts = ["This is the first document.", "This is the second document.", "This is compute."]
)
```
